```python
import pandas as pd
import numpy as np

path = 'data/bank.csv'

# engine='python' и sep=None заставляют Pandas самому понять, какой разделитель в файле
df = pd.read_csv(path, sep=None, engine='python')

# Убираем пробелы и лишние символы из названий колонок
df.columns = df.columns.str.strip().str.replace('"', '')

categorical_cols = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome']
for col in categorical_cols:
    df[col] = df[col].astype(str).str.replace('.', ' ', regex=False).str.title()
    df[col] = df[col].replace('Unknown', 'Not Specified')

df['conversion_status'] = np.where(df['deposit'] == 'yes', 1, 0)

age_bins = [0, 25, 40, 60, 120]
age_labels = ['18-25', '26-40', '41-60', '60+']
df['age_segment'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)

df['duration_min'] = round(df['duration'] / 60, 2)

headers = {
    'job': 'Occupation',
    'marital': 'Marital_Status',
    'education': 'Education_Level',
    'housing': 'Housing_Loan',
    'loan': 'Personal_Loan',
    'balance': 'Balance',
    'month': 'Month',
    'poutcome': 'Prev_Outcome',
    'deposit': 'Subscribed'
}

df_final = df.rename(columns=headers)
df_final.to_csv('bank_for_looker.csv', index=False)

print("Done")
```
